## Imports

```python
import pandas as pd
from sqlalchemy import create_engine, text
```

## Database connection

```python
engine = create_engine(
    "mysql+pymysql://root:password@localhost:3306/classicmodels"
)
```

```python
try:
    with engine.connect() as connection:
        print("Connected successfully!")
except Exception as error:
    print("Connection failed:")
    print(error)
```

## Products table

```python
products_query = """
SELECT *
FROM products;
"""

with engine.connect() as connection:
    products_df = pd.read_sql(
        text(products_query),
        connection
    )

print(products_df.head())
```

## 1. First 10 records

```python
print(products_df.head(10))
```

## 2. Product code and product name

```python
print(products_df[['productCode', 'productName']].head(10))
```

## 3. Set productCode as the index

```python
with engine.connect() as connection:
    products_indexed_df = pd.read_sql(
        text(products_query),
        connection,
        index_col="productCode"
    )

print(products_indexed_df.head())
```

## 4. Basic statistics

```python
print("\nBasic Statistics:")
print(products_indexed_df.describe())
```

## 5. Data types

```python
print("\nData Types:")
print(products_indexed_df.dtypes)
```

## 6. Rows and columns

```python
print("\nRows and Columns:")
print(products_indexed_df.shape)

print("\nNumber of Rows:")
print(products_indexed_df.shape[0])

print("\nNumber of Columns:")
print(products_indexed_df.shape[1])
```

## 7. Missing values

```python
print("\nMissing Values:")
print(products_indexed_df.isnull().sum())
```

## 8. Grouping and aggregation

```python
grouped_products = (
    products_indexed_df
    .groupby("productLine")
    .agg(
        totalQuantityInStock=("quantityInStock", "sum"),
        averageBuyPrice=("buyPrice", "mean")
    )
    .reset_index()
)

grouped_products["averageBuyPrice"] = (
    grouped_products["averageBuyPrice"].round(2)
)

print("\nGrouped Product Data:")
print(grouped_products)
```

## 9. Order details

```python
orders_query = """
SELECT
    orderNumber,
    productCode,
    priceEach,
    orderLineNumber,
    quantityOrdered
FROM orderdetails;
"""

with engine.connect() as connection:
    orders_df = pd.read_sql(
        text(orders_query),
        connection
    )

print("\nSample Order Details:")
print(orders_df.head())
```

### Total cost for each order line

```python
orders_df["totalCost"] = (
    orders_df["priceEach"]
    * orders_df["quantityOrdered"]
)

print(orders_df.head())
```

### Total amount for each order

```python
order_totals = (
    orders_df
    .groupby("orderNumber", as_index=False)["totalCost"]
    .sum()
)

order_totals["totalCost"] = (
    order_totals["totalCost"].round(2)
)

print("\nTotal Amount for Each Order:")
print(order_totals)
```
